```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('./auto-mpg.csv')

# 열이름 지정
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight',
              'acceleration', 'model year', 'origin', 'name']

df.count()
```

```python
type(df.count())
```

```python
# 데이터 개수 확인
# 데이터 프레임 df의 특정 열이 가지고 있는 고유값의 개수 확인
unique_values = df['origin'].value_counts()
unique_values
```

```python
# 데이터 프레임 df의 특정 열이 가지고있는 고유값의 비율을 확인
unique_values_ratio = df['origin'].value_counts(normalize=True)
unique_values_ratio
```

```python
# 데이터 프레임 df의 특정 열이 가지고 있는 고유값의 비율을 확인(백분율로 변환)
unique_values_percentage = (df['origin'].value_counts(normalize=True) * 100).round(1)
unique_values_percentage
```

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('./auto-mpg.csv')

# 열이름 지정
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight',
              'acceleration', 'model year', 'origin', 'name']

# 평균값
print(df.mean(numeric_only=True))
print('\n')
print(df['mpg'].mean())
print(df.mpg.mean())
print('\n')
print(df[['mpg', 'weight']].mean())
```

```python
# 중앙값
print(df.median(numeric_only=True))
print('\n')
print(df['mpg'].median())
```

```python
# 최대값
print(df.max(numeric_only=True))
print()
print(df['mpg'].max())
```

```python
# 최솟값
print(df.min(numeric_only=True))
print()
print(df.min())# 문자열로도 그냥 값 카운트해버림
print()
print(df['mpg'].min())
```

```python
# 표준편차
print(df.std(numeric_only=True))
print()
print(df['mpg'].std())
```

```python
print(df.corr(numeric_only=True))
print()
print(df[['mpg', 'weight']].corr())
```

```python
import pandas as pd

# 데이터프레임 변환
df = pd.read_excel('../../eda/data/남북한발전전력량.xlsx')

# 남한, 북한 발전량 합계 데이터만 추출
df_ns = df.iloc[[0,5], 3:]

# 행 인덱스 변경
df_ns.index = ['South', 'North']

# 발전량 데이터의 자료형을 문자열에서 정수형으로 변환
df_ns = df_ns.astype(int)

# 행, 열 전치하여 막대 그래프 그리기
tdf_ns = df_ns.T
print(tdf_ns.head())
print()
tdf_ns.plot(kind='bar')
```

```python
# 누적 막대 그래프
tdf_ns.plot(kind='bar', stacked=True) # bar플랏으로 찍고 쌓아서 찍어라
```

```python
import pandas as pd

# 데이터프레임 변환
df = pd.read_excel('../../eda/data/남북한발전전력량.xlsx')

# 남한, 북한 발전량 합계 데이터만 추출
df_ns = df.iloc[[0,5], 3:]

# 행 인덱스 변경
df_ns.index = ["South", "North"]

# 발전량 데이터의 자료형을 문자열에서 정수형으로 변환
df_ns = df_ns.astype(int)

# 행, 열 전치하여 히스토그램 그리기
tdf_ns = df_ns.T
tdf_ns.plot(kind='hist')
```

```python
tdf_ns.plot(kind='hist',
            alpha=0.5,
            figsize=(6,2),
            grid=True,
            color='skyblue',
            edgecolor='black')
```

```python
# hist 메서드 활용
tdf_ns.hist(figsize=(6,2))
```

```python
# 차트 서식 설정
tdf_ns.hist(bins=20, grid=False, figsize=(6,2))
```
