# p 태그 문단 가져오기
- html에서 p태그는 문단을 의미
- find_all()은 조건에 맞는 태그를 모두 찾아 리스트처럼 변환


```python
paragraphs = soup.find_all('p')
print(paragraphs[0].text)
print(len(paragraphs))


for i in range(3):
    print(paragraphs[i])
```

```python
paragraphs_list = []

# 모든 p 태그를 순회하면서 하나씩 반복하는데 
# 문자열의 앞뒤 공백을 없애고 줄바꿈 등을 제거
for p in paragraphs:
    text = p.text.strip()

    # 빈 문단을 제거하고 리스트에 넣어
    if text != "":
        paragraphs_list.append(text)


# 리스트를 데이터 프레임으로 변환
paragraph_df = pd.DataFrame(paragraphs_list, columns=['문단'])

# 상위 5개 행 확인
print(paragraph_df.head())
```

# a태그에서 텍스트랑 링크 정보 가져오기
- HTML 에서 a태그는 링크를 의미
- link.text: 링크가 걸린 글자를 의미
- link.get('속성명 ex) href'): 링크 주소

```python
links = soup.find_all('a')

for link in links:
    print('글자', link.text.strip())
    print(link.get('href'))
```

```python
# 링크 보를 데이터프레임으로 저장
link_data = []

for link in links:
    # 링크 걸린 글자만 공백 제거한 후 텍스트에 저장
    text = link.text.strip()

    # href 속성에 들어있는 링크 주소만 href의 변수에 저장
    href = link.get("href")
    
    # 결측 값이 없는 경우에만 데이터 프레임에 저장한다.
    if text != "" and href is not None:
        link_data.append([text, href])

# 리스트를 데이터 프레임으로 변환
pandas_dataframe = pd.DataFrame(link_data, columns=['글자', '링크'])
```

```python
# 라이브러리 불러오기
import requests
import pandas as pd
from dotenv import load_dotenv
import os

load_dotenv()

KAKAO_MAP_KEY = os.getenv('KAKAO_MAP_KEY')
```

```python
def get_geocoding(place):
    url = 'https://dapi.kakao.com/v2/local/search/address.json'

    headers = {
        'Authorization': f"KakaoAK {KAKAO_MAP_KEY}"
    }

    # place 는 함수에 전달받은 주소를 저장하는 변수
    # 카카오 API의 query 값으로 전달되어 해당 주소의 위도와 경도를 검색
    params = {
        'query': place
    }

    response = requests.get(url, headers=headers, params=params)
    result = response.json()
    return result['documents'][0]['y'], result['documents'][0]['x']
```

```python
lat = [] # 위도
lng = [] # 경도

# 장소(주소) 리스트
places = ['서울특별시 종로구 세종대로 175',
          '서울특별시 서초구 서초동 700',
          '부산광역시 해운대구 해운대해변로 264']

i = 0
for place in places:
    i += 1
    try:
        print(i, place)
        place_lat, place_lng = get_geocoding(place)
        lat.append(place_lat) # 위도를 리스트에 추가
        lng.append(place_lng) # 경도를 리스트에 추가
    except Exception as e:
        print('오류 발생', e)
        lat.append('')
        lng.append('')

df = pd.DataFrame({'위도': lat, '경도': lng}, index=places)
df
```
