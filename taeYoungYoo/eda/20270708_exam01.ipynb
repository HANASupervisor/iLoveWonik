```python
import pandas as pd

# 딕셔너리 데이터로 판다스 시리즈 만들기
student1 = pd.Series({'국어': 100, '영어': 80, '수학': 90})
student2 = pd.Series({'수학': 80, '국어': 90, '영어': 80})

print(student1)
print()
print(student2)

# 두학생의 과목별 점수로 사칙연산 수행
addition = student1 + student2
print(type(addition))

print(addition)
```

```python
subtraction = student1 - student2
print(type(subtraction))
print(subtraction)
```

```python
multiplication = student1 * student2
print(type(multiplication))
print(multiplication)
```

```python
division = student1 / student2
print(type(division))
print(division)
```

```python
# 사칙연산 결과를 데이터 프레임으로 합치기(시리즈 -> 데이터프레임)
result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index = ['덧셈', '뺄셈', '곱셈', '나눗셈'])

print(result)
```

```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np

# 딕셔너리 데이터로 판다스 시리즈 만들기
student1 = pd.Series({'국어': np.nan, '영어': 80, '수학': 90})
student2 = pd.Series({'수학': 80, '국어': 90})

print(student1)
print(student2)

addition = student1 + student2
subtraction = student1 - student2
multiplication = student1 * student2
division = student1 / student2
print(type(division))
print()

# 사칙연산 결과를 데이터프레임으로 합치기(시리즈 -> 데이터 프레임)
result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index=['덧셈', '뺄셈', '곱셈', '나눗셈'])

print(result)
```

```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np

# 딕셔너리 데이터로 판다스 시리즈 만들기
student1 = pd.Series({'국어': np.nan, '영어': 80, '수학': 90})
student2 = pd.Series({'수학': 80, '국어': 90})

print(student1)
print(student2)

# 두 학생의 과목별 점수로 사칙연산 수행(연산 메서드 사용)
sr_add = student1.add(student2, fill_value=0)
sr_sub = student1.sub(student2, fill_value=0)
sr_mul = student1.mul(student2, fill_value=0)
sr_div = student1.div(student2, fill_value=0)

result = pd.DataFrame([sr_add, sr_sub, sr_mul, sr_div],
                      index=['덧셈', '뺄셈', '곱셈', '나눗셈'])

print(result)
```

```python
import pandas as pd
import seaborn as sns 

# titanic 데이터셋에서 age, fare 2개 열ㅇ르 선택하여 데이터 프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age', 'fare']]
print(df.tail())
print()
print(type(df))
print()

# 데이터프레임에 숫자 10 더하기
addition = df + 10
print(addition.tail())
print()
print(type(addition))
print()

# 데이터프레임끼리 연산하기(addition - df)
subtraction = addition - df
print(subtraction.tail())
print()
print(type(subtraction))
```

```python
# 사칙연산 수행(연산 메서드 사용)
sample1 = addition.tail()
sample2 = addition.tail().fillna(0.0)

print(sample1)
print(sample2)
```
