```python
import pandas as pd

# 딕셔너리를 데이터프레임으로 변환
dict_data = {'c0': [1, 2, 3], 'c1':[4, 5, 6], 'c2': [7, 8, 9], 'c3': [10, 11, 12], 'c4': [13, 14, 15]}

# 인덱스를 {r0, r1, r2}로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)

# 인덱스를 [r0, r1, r2, r3, r4]로 재지정
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf = df.reindex(new_index)
print(ndf)

# reindex로 발생한 NaN 값을 숫자 0으로 채우기
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf = df.reindex(new_index, fill_value=0)
print(ndf)
```

```python
import pandas as pd

# 딕셔너리 데이터로 판다스 시리즈 만들기
student1 = pd.Series({'국어': 100, '영어': 80, '수학': 90})
student2 = pd.Series({'수학': 80, '국어': 90, '영어': 80})

print(student1)
print()
print(student2)

# 두학생의 과목별 점수로 사칙연산 수행
addition = student1 + student2
print(type(addition))

print(addition)
```

```python
subtraction = student1 - student2
print(type(subtraction))
print(subtraction)
```

```python
multiplication = student1 * student2
print(type(multiplication))
print(multiplication)
```

```python
division = student1 / student2
print(type(division))
print(division)
```

```python
# 사칙연산 결과를 데이터 프레임으로 합치기(시리즈 -> 데이터프레임)
result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index = ['덧셈', '뺄셈', '곱셈', '나눗셈'])

print(result)
```

```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np

# 딕셔너리 데이터로 판다스 시리즈 만들기
student1 = pd.Series({'국어': np.nan, '영어': 80, '수학': 90})
student2 = pd.Series({'수학': 80, '국어': 90})

print(student1)
print(student2)

addition = student1 + student2
subtraction = student1 - student2
multiplication = student1 * student2
division = student1 / student2
print(type(division))
print()

# 사칙연산 결과를 데이터프레임으로 합치기(시리즈 -> 데이터 프레임)
result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index=['덧셈', '뺄셈', '곱셈', '나눗셈'])

print(result)
```

```python
import pandas as pd
import seaborn as sns

# titanic  데이터셋에서 age, fare 2개 열을 선택하여 데이터 프레임 만들기
titanic = sns.load_dataset('titanic')
print(titanic)
df = titanic.loc[:, ['age', 'fare']]
print(df.head())
print()
print(type(df))
```

```python
# 데이터프레임에 숫자 10 더하기
addition = df + 10
print(addition.head())
print()
print(type(addition))
```

```python
# 데이터프레임까지 연산하기(addition -df)
subtraction = addition - df
print(subtraction.tail())
print()
print(type(subtraction))
```
