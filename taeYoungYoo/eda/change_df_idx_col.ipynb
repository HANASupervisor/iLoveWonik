## 데이터 프레임의 행 데이터를 선택하기 위해서는 loc과 iloc 인덱서를 사용한다. 인덱스 이름을 기준으로 행을 선택할 때는 loc을 이용하고, 정수 위치 인덱스를 사용할때는 iloc을 이용한다.

```python
label1 = df.loc['서준']
position1 = df.iloc[0]
print(label1)
print()
print(position1)
```

```python
# 데이터 프레임에서 여러 행 선택
label2 = df.loc[['서준', '우현']]
position2 = df.iloc[[0, 1]]
print(label2)
print()
print(position2)
```

```python
# 열조회
print(df.loc[:, '수학'])
print(df.iloc[:, 0])
```
