```python
import pandas as pd
import seaborn as sns

# titanic  데이터셋 로딩
titanic = sns.load_dataset('titanic')

# 행 인덱스 0~9 범위에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
df = titanic.loc[0:9, ['age', 'fare']]

df
```

```python
# 조건식 적용(20세 미만: age < 20)
df['age'] < 20
```

```python
# 불린 인덱싱(20세 미만의 조건을 충족하는 행을 필터링)
df[df['age'] < 20]
```

```python
# 불린 인덱싱(20세 미만의 조건을 충족하는 행을 필터링)
df.loc[df['age'] < 20]['age']
```

```python
df.loc[df['age'] < 20, 'age']
```

```python
df.loc[~(df['age'] < 20)]
```

```python
# 나이가 10대(10~19세)인 승객(AND 논리연산)
mask1 = (titanic.age >= 10) & (titanic.age < 20)
df_teenage = titanic[mask1]
df_teenage.head()
```

```python
mask2 = (titanic.age < 10) & (titanic.sex =='female')
df_female_under10 = titanic.loc[mask2]
df_female_under10.head()
```

```python
# 나이가 10세 미만(0~9세) 또는 60세 이상인 승객 (OR  논리연산)
# age, sex, alone 열만 선택 (loc 인덱서, fancy indexing)
mask3 = (titanic.age < 10) | (titanic.age >=60)
df_under10_morethan60 = titanic.loc[mask3, ['age', 'sex', 'alone']]
print(df_under10_morethan60.head())
```

# 문자열 메서드 사용하기

```python
import pandas as pd
import numpy as np
```

```python
# 텍스트로 이루어진 시리즈 배열 만들기(자료형 지정, 일반 객체 Object 생성)
ser = pd.Series(['Apple_사과', 'Banana_바나나', 'Cherry_체리', np.nan],
                index=['First', 'Second', 'Third', 'Fourth'])
print(ser)
```

```python
# 텍스트로 이루어진 시리즈 배열 만들기(자료형 지정, 일반 객체 Object 생성)
ser = pd.Series(['Apple_사과', 'Banana_바나나', 'Cherry_체리', np.nan],
                index=['First', ' Second', 'Third ', 'Fourth '], dtype='string')
print(ser)
```

```python
print(ser.str.lower())
print(ser)
```

```python
print(ser.str.len())
```

```python
# 문자열 분할
print(ser.str.split('_'))
print()

# 문자열을 분할하여 DataFrame으로 확장
print(ser.str.split('_', expand=True))
print()

# 분할된 문자열 중 특정 요소 선택
print(ser.str.split('_').str.get(1))
```

```python
# 정규식 표현 찾아서 정리해라.
```

```python
print(ser.str.replace("_", ":", regex=False))
```

```python
# replace()는 문자열 교체
# regex = False 일반 문자 그대로 찾아라
# regex = True 정규식 표현으로 찾아라
# 영어 소문자, 대문자, 공백을 제외한 문자를 제거한다.
print(ser.str.replace("[a-zA-Z\s_]", "", regex=True))
```

```python
# r은 raw 임 왜 이걸쓰냐면 regex상황에 바로 넣어야함
# escape문이 들어가는 경우 이슈가 발생할수 있음. 그래서 그냥 문자열 그대로 정규식에 넣기위해 r을 활용해서 작성함.
# 첫글자가 A또는 B로 시작하고 그 뒤에 소문자가 하나이상 있는 문자열 검색
# 그리고 정규표현식은 후위표기법으로 작성해야함.

print(ser.str.replace(r"[^가-힣]", "", regex=True))
```

```python
print(ser.str.contains(r"[A|B][a-z]+"))
```

```python
# fullmatch() 전체 문자열의 패턴 일치 여부 확인
# contains()는 문자열이 포함되어있으면 True
# fullmatch()는 전체가 정확하게 같으면 True
# 첫글자가 A 또는 B로 시작하고, 그 뒤에 소문자가 하나이상인 경우 True
print(ser.str.fullmatch(r"[AB][a-z]+"))
```
