```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from io import StringIO # 라이브러리 이름 생략하고 불러올 수 있다
```

```python
# 스크래핑할 웹페이지 주소
url = 'https://ko.wikipedia.org/wiki/파이썬'

# 일부 웹사이트는 파이썬 프로그램의 접근을 막을 수 있으므로 User-Agent 를 지정
headers = {
    'User-Agent': 'Mozilla/5.0'
}

# 웹페이지 요청
response = requests.get(url, headers = headers)

# 응답 상태 코드 확인
# 200이면 정상적으로 가져온 것
print(response.status_code)
```

```python
# print(response.text)
```

```python
# HTML 문서를 BeautifulSoup 으로 객체 변환
soup = BeautifulSoup(response.text,'lxml')
# 문서 제목 확인
print(soup.title.text)
```

# p태크에서 문단 가져오기
- HTML에서 P 태그는 문단을 의미
- find_all()은 조건에 맞는 태그를 모두 찾아 리스트처럼 반환

```python
# 모든 p 태그 찾기
paragraphs = soup.find_all('p')
print(paragraphs[1])
```

```python
# 찾은 p 태그의 개수 구하기
print(len(paragraphs))
```

```python
# 앞의 3개 문단만 출력
for p in paragraphs[:3]:
    text = p.text.strip()

    # 빈 문단을 제외하고 출력함
    if text != '':
        print(text)
        print('-'*60)
```

```python
# 문자 내용을 저장할 빈 리스트 생성
paragraph_data = []

# 모든 p 태그를 순회하면서 하나씩 반복
# strip() 메서드로 문자열의 앞 뒤 공백, 탭, 줄 바꿈 등을 제거

for p in paragraphs:
    text = p.text.strip()

    # 빈 문단을 제외
    if text != '':
        paragraph_data.append(text)

# 리스트를 Dataframe으로 변환
paragraph_df = pd.DataFrame(paragraph_data, columns=['문단'])

# 상위 5개 행 확인
paragraph_df.head()
```

# a 태그에서 텍스트와 링크 정보 가져오기
- HTML 에서 a 태그는 링크를 의미
- link.text: 링크 걸린 글자
- link.get('href') : 링크 주소

```python
# 모든 a 태그 찾기
links = soup.find_all('a')
print(len(links))

# 앞의 10개 링크의 텍스트와 링크 주소 출력
for link in links[:10]:
    print('글자:',link.text.strip())
    print('주소:',link.get('href'))
    print('-'*40)
```

```python
# 링크 정보를 DataFrame으로 저장
link_data = []

for link in links:
    # 링크 걸린 글자만 공백 제거한 후 text에 저장
    text = link.text.strip()

    # href 속성에 들어있는 링크 주소만 href 변수에 저장
    href = link.get('href')

    # 글자와 주소가 모두 있는 경우만 리스트에 저장
    if text != '' and href is not None:
        link_data.append([text,href])

# 리스트를 DataFrame으로 변환
link_df = pd.DataFrame(link_data, columns=['글자','링크'])

link_df.head()
```
