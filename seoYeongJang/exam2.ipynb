# HTML 파일 불러오기
- read_html() 함수는 HTML 웹페이지에 있는 태그를 표 형식의 데이터로 모두 찾아서 데이터프레임으로 반환
- 표가 아닌 다른 HTML 요소는 Dataframe 으로 반환하지 못함

```python
import pandas as pd

tables = pd.read_html('../data/sample.html')
print(tables)
```

```python
# 표(table)의 개수 확인
print(len(tables))
```

```python
# tables 리스트의 원소를 반복하면서 각각 출력
for i in range(len(tables)):
    print(tables[i])
    print('\n')
```

```python
url= 'https://en.wikipedia.org/wiki/Korea'

import requests 
from io import StringIO # 메모리에 있는 문자열을 파일처럼 사용할 수 있게 함

headers = {
    'User-Agent': 'Mozilla/5.0'
}

response = requests.get(url, headers = headers)
##############################################
# HTML 문자열을 StringIO 객체로 변환
# StringIO는 메모리에 있는 문자열을 파일처럼 읽을 수 있게 만들어주는 객체
tables = pd.read_html(StringIO(response.text))

print(len(tables))
```

```python
tables[0]
```

```python
for i in range(len(tables)):
    print(tables[i])
    print('\n')
```
