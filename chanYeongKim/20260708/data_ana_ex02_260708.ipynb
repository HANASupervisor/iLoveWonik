```python
import pandas as pd
from numpy.testing import print_coercion_tables
```

```python
exam_data = {'이름': ['서준', '우현', '인아'],
             '수학': [ 90, 80, 70],
             '영어': [ 98, 89, 95],
             '음악': [ 85, 95, 100],
             '체육': [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
```

```python
# '이름' 열을 새로운 인덱스로 지정하고 df 객체에 변경 사항 저장
df = df.set_index('이름')
print(df)
```

```python
df.iloc[0] [3] = 80
print(df)
```

```python
# iloc의 경우 현재 버전에서는 행과 열을 한 번에 지정해야 함
df.iloc[0, 3]= 80
print(df)
```

```python
df.loc['서준']['체육'] = 90
print(df)
```

```python
df.loc['서준', '체육'] = 90
print(df)
```

```python
# 여러 개의 원소 값을 수정하는 방법
df.loc['서준', ['음악', '체육']] = 90
print(df)
```

```python
# 여러 개의 원소 값을 수정하는 방법
df.loc['서준', ['음악', '체육']] = 90, 50
print(df)
```

```python
df.loc[['서준', '우현'], ['음악', '체육']] = 90, 50
print(df)
```

```python
df.loc[['서준', '우현'], ['음악', '체육']] = [[90, 80], [70,60]]
print(df)
```

```python
## 전치, 데이터프레임의 행과 열을 서로 맞바꾸는 방법
exam_data = {'이름': ['서준', '우현', '인아'],
             '수학': [ 90, 80, 70],
             '영어': [ 98, 89, 95],
             '음악': [ 85, 95, 100],
             '체육': [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
```

```python
# 데이터프레임의 transpose() 메소드로 전치하기
df = df.transpose()
print(df)
```

```python
# 데이터프레임의 T 속성으로 전치하기
df =df.T
print(df)
```

```python
## 특정 열을 행 인택스로 지정
ndf = df.set_index(['이름'])
print(ndf)
```

```python
# 하나의 열을 인덱스로 지정
ndf2 = ndf.set_index('음악')
print(ndf2)
```

```python
# 두 개의 열을 인덱스로 지정
ndf3 = ndf.set_index(['수학', '음악'])
print(ndf3)
```

```python
# 딕셔너리를 정의
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 딕셔너리를 데이터프레임으로 변환, 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
```

```python
# 인덱스로 사용할 새로운 리스트 생성
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf = df.reindex(new_index)
print(ndf)
```

```python
# reindex로 발생한 NaN값을 숫자 0으로 채우기
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf2= df.reindex(new_index, fill_value=0)
print(ndf2)
```

```python
# 딕셔너리를 정의
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,11], 'c4':[14,14,15]}

# 딕셔너리를 데이터프레임으로 변환, 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
```

```python
# c1 열을 기준으로 내림차순 정렬
ndf = df.sort_values(by='c1', ascending=False)
print(ndf)
```

```python
# c2 열을 기준으로 오름차순 정렬
ndf2 = df.sort_values(by='c2')
print(ndf2)
```

```python
# c3 열을 기준으로 내림차순 정렬, c4 기준으로 오름차순 정렬
ndf3 = df.sort_values(by=['c3', 'c4'], ascending=[False, True])
print(ndf3)
```
