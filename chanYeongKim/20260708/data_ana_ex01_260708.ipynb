# 시리즈와 데이터프레임의 차이
  - 시리즈는 딕셔너리 한개를 세로로 펼친 것, 키가 인덱스가 됨
  - 데이터프레임은 여러 개의 리스트를 표로 만든 것, 키가 열이 됨

```python
exam_data={'이름':'김찬영',
           '수학':80,
           '영어':100,
           '국어':90
           }
```

```python
print(exam_data)
```

```python
import pandas as pd
sr = pd.Series(exam_data)
print(sr)
```

```python
 # # 시리즈 하나로 데이터프레임 만들려면 오류뜸
 # df = pd.DataFrame(exam_data)
```

```python
exam_data2 = {'이름':['김찬영','홍길동', '성춘향'],
              '수학':[80,90,100],
              '영어':[80,90,100],
              '국어':[80,90,100],
              }
df = pd.DataFrame(exam_data2)
print(df)
```

```python
exam_data = {'이름': ['서준', '우현', '인아'],
             '수학': [ 90, 80, 70],
             '영어': [ 98, 89, 95],
             '음악': [ 85, 95, 100],
             '체육': [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
```

```python
print(type(df))
```

```python
df = df.set_index('이름')
print(df)
```

```python
print(df['수학'])
```

```python
print(df['영어'])
```

```python
# # 데이터프레임 인덱싱은 column만 가능 row 안됨
# print(df['서준'])
```

```python
print(df[['수학', '영어']])
```

```python
# column 생략 가능
print(df.loc['서준'])
print(df.loc['서준',:])
```

```python
# 행을 선택할 때는 loc 인덱서를 사용
# df.loc[행이름, 열이름]
print(df.loc['서준', '음악'])
```

```python
# column만 쓰는 것은 안됨. 전체 출력하고 싶으면 열 앞에 :, 사용
print(df.loc[:,'음악'])
```

```python
exam_data = {'이름': ['서준', '우현', '인아'],
             '수학': [ 90, 80, 70],
             '영어': [ 98, 89, 95],
             '음악': [ 85, 95, 100],
             '체육': [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
```

```python
#새로운 행을 추가
df.loc[3] = 0
print(df)
```

```python
df.loc[4] = ['동규', 90, 80, 70, 60]
print(df)
```

```python
# 기존의 행을 복사해서 새로운 행을 만드는 방법
df.loc['행5'] = df.loc[3]
print(df)
```

```python
df.loc['행6'] = df.loc['행5']
print(df)
```

```python
df.loc['행7'] = df.iloc[6]
print(df)
```
